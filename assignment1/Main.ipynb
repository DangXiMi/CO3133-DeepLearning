# Assignment 1: Train/Validation loop for the linear classifier and the MLP


```python
import sys
from pathlib import Path

REPO_ROOT = Path.cwd().resolve()
sys.path.insert(0, str(REPO_ROOT / 'assignment1'))
sys.path.insert(0, str(REPO_ROOT))

from training_loop import Linear, MLP, get_dataloaders, load_split, preprocess, set_seed, train
```

## 1. Dataset, split and DataLoaders

```python
set_seed()

(X_train, y_train), (X_validate, y_validate), (X_test, y_test) = load_split(dataset_name='fashion_mnist')
(X_train, X_validate, X_test), (mean, std) = preprocess(X_train, X_validate, X_test)
train_loader, validation_loader, test_loader = get_dataloaders(X_train, y_train, X_validate, y_validate, X_test, y_test)

print('Train/Validation/Test:', len(y_train), len(y_validate), len(y_test))
print('input shape and dtype:', tuple(X_train.shape), X_train.dtype)
```

## 2. Train the linear classifier and the MLP


```python
for name, build in [('Linear', Linear), ('MLP', MLP)]:
    set_seed()
    model = build()
    print(f'{name} ({model.num_para()} parameters)')
    train(model, train_loader, validation_loader)
```
